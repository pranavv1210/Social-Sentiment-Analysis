```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from wordcloud import WordCloud

# Load cleaned dataset
df = pd.read_csv("../data/sentiment_tweets.csv")

# Sentiment distribution
sns.countplot(x="airline_sentiment", data=df)
plt.title("Sentiment Distribution")
plt.show()

# WordClouds for each sentiment
for sentiment in df["airline_sentiment"].unique():
    text = " ".join(df[df["airline_sentiment"] == sentiment]["clean_text"].astype(str))
    wc = WordCloud(width=800, height=400, background_color="white").generate(text)
    plt.figure(figsize=(10,6))
    plt.imshow(wc, interpolation="bilinear")
    plt.axis("off")
    plt.title(f"WordCloud for {sentiment} tweets")
    plt.show()
```
